# Step 8: Apply All Your Imputation Decisions

```python
df_clean = df.copy()

# 8a. Impute bedrooms & beds using median-by-accommodates group
df_clean['bedrooms'] = df_clean.groupby('accommodates')['bedrooms'] \
    .transform(lambda x: x.fillna(x.median()))
df_clean['beds'] = df_clean.groupby('accommodates')['beds'] \
    .transform(lambda x: x.fillna(x.median()))

# Fallback: if still missing (rare accommodates group with no data), use overall median
df_clean['bedrooms'] = df_clean['bedrooms'].fillna(df_clean['bedrooms'].median())
df_clean['beds'] = df_clean['beds'].fillna(df_clean['beds'].median())

# 8b. Fill reviews_per_month with 0 where number_of_reviews == 0
df_clean.loc[df_clean['number_of_reviews'] == 0, 'reviews_per_month'] = \
    df_clean.loc[df_clean['number_of_reviews'] == 0, 'reviews_per_month'].fillna(0)

# 8c. review_scores_rating: create a flag instead of guessing a fake score
df_clean['has_reviews'] = (df_clean['number_of_reviews'] > 0).astype(int)

print("=== AFTER IMPUTATION — remaining missing in key columns ===")
for c in ['bedrooms', 'beds', 'reviews_per_month', 'review_scores_rating']:
    print(f"{c}: {df_clean[c].isna().sum()} missing")
```

Fills missing bedrooms/beds using group median, fills reviews_per_month
with 0, and keeps review_scores_rating missing (adds a has_reviews flag
instead of faking a score).

# Markdown: boundary justification

### Deciding price_category boundaries

Two options considered:
1. **Domain-based cutoffs** (e.g. Low < $100, Medium $100–300, High > $300) — interpretable,
   matches how a host/guest would naturally think about price tiers.
2. **Quantile-based (tertiles)** — guarantees roughly balanced classes for modeling, but
   boundaries are data-dependent and less intuitive.

**Chosen: Domain-based cutoffs — Low < $100, Medium $100–300, High > $300.**

This is based on the IQR analysis above: Q1 = $96.88 and Q3 = $302.72, meaning the
"Medium" band (roughly $100–300) already captures the bulk (~50%) of listings by
construction, since it approximates the interquartile range itself. Rounding Q1 and Q3
to $100 and $300 keeps the category boundaries clean and explainable to a non-technical
reader, while still reflecting the actual shape of the price distribution rather than
being arbitrary.

The IQR bounds ([-211.89, 611.49]) confirm the lower bound is meaningless (price can't be
negative), so no listings are flagged as low-price outliers — all 1,608 outliers (7.47%
of data) are high-price extremes above $611.49.

# Step 6 - Create the target variable (Preprocessing)

```python
bins = [0, 100, 300, np.inf]
labels = ['Low', 'Medium', 'High']
df['price_category'] = pd.cut(df['price'], bins=bins, labels=labels)

print(df['price_category'].value_counts())
```

# Step 7 - Outlier treatment decision (Preprocessing)

```python
# Decision: CAP (winsorize) upper extreme prices rather than drop them.
# Lower bound (-211.89) is meaningless since price can't be negative, so only
# the upper bound (611.49) is applied.

upper_bound = 611.49  # from Cell 4

df['price_capped'] = df['price'].clip(upper=upper_bound)

# Recreate the category from the capped price to keep it consistent:
df['price_category'] = pd.cut(df['price_capped'], bins=[0, 100, 300, np.inf], labels=['Low', 'Medium', 'High'])

print(f"Rows affected by capping: {(df['price'] != df['price_capped']).sum()}")
print(df['price_category'].value_counts())
```

Justification: dropping ~1,608 rows (7.47% of data) would lose signal and shrink the
training set unnecessarily. Capping preserves every row while preventing extreme high
prices from dominating distance-based or gradient-based models downstream. Since no
listings fall below the lower bound, only high-end prices are capped.

Preprocessing
Cell 11: Choose features and exclude leakage

```python
LEAKAGE_COLS = ['price', 'price_capped']

categorical_features = [c for c in [
    'room_type', 'neighbourhood_group_cleansed', 'neighbourhood_cleansed', 'property_type',
    'host_response_time', 'host_is_superhost', 'host_identity_verified',
    'host_has_profile_pic', 'has_availability', 'instant_bookable'] if c in df.columns]

numeric_features = [c for c in df.select_dtypes(include='number').columns
                    if c not in LEAKAGE_COLS + [TARGET]]

excluded = [c for c in df.columns
            if c not in categorical_features + numeric_features + LEAKAGE_COLS + [TARGET]]

print("Categorical features:", categorical_features)
print("\nNumeric features:", numeric_features)
print("\nLeft out (text / dates / high-cardinality etc.):", excluded)
```

Cell 12: Train/test split

```python
from sklearn.model_selection import train_test_split

print("Rows with missing target:", df[TARGET].isna().sum())   # should be 0
model_df = df.dropna(subset=[TARGET])

X = model_df[categorical_features + numeric_features]
y = model_df[TARGET]

assert not any(c in X.columns for c in LEAKAGE_COLS), "Leakage column still in X!"

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print("Train:", X_train.shape, " Test:", X_test.shape)
```

Cell 13: Encode categoricals and scale numerics

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),      # Logistic Regression / KNN cannot handle NaN
    ('scaler', StandardScaler())                        # KNN and Logistic Regression need scaled features
])

categorical_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='infrequent_if_exist',
                             min_frequency=30,          # categories with under 30 train rows are grouped as "infrequent"
                             sparse_output=False))
])

# Identify all columns that are entirely empty in the current DataFrame
all_fully_empty_cols = df.columns[df.isna().all()].tolist()

# Filter out these problematic columns from both lists
# We make copies of the lists to avoid modifying the originals from the previous cell directly
numeric_features_filtered = [
    c for c in numeric_features if c not in all_fully_empty_cols
]
categorical_features_filtered = [
    c for c in categorical_features if c not in all_fully_empty_cols
]

preprocessor = ColumnTransformer([
    ('num', numeric_pipe, numeric_features_filtered),
    ('cat', categorical_pipe, categorical_features_filtered)
])
preprocessor.set_output(transform='pandas')

X_train_enc = preprocessor.fit_transform(X_train)   # learn from TRAIN only
X_test_enc = preprocessor.transform(X_test)         # apply to test

print("Encoded train:", X_train_enc.shape, " test:", X_test_enc.shape)
print("Any NaN left in train:", X_train_enc.isna().sum().sum())
X_train_enc.iloc[:, :8].describe().T[['mean', 'std']].round(2)
```

Cell 14: Feature selection

```python
# Filter 1a: near-constant columns
low_var = X_train_enc.columns[X_train_enc.var() < 0.001].tolist()
print("Near-constant columns removed:", low_var)

# Filter 1b: highly correlated numeric pairs (e.g. availability_30 vs availability_60)
num_enc_cols = [c for c in X_train_enc.columns if c.startswith('num__')]
corr_matrix = X_train_enc[num_enc_cols].corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
pairs = [(a, b, round(upper.loc[a, b], 3)) for a in upper.index for b in upper.columns
         if pd.notna(upper.loc[a, b]) and upper.loc[a, b] > 0.90]
print("\nPairs with |corr| > 0.90:")
for a, b, r in pairs:
    print(f"  {a}  <->  {b}   r={r}")
to_drop_corr = sorted({b for a, b, r in pairs})
print("Dropping:", to_drop_corr)

# Filter 2: mutual information ranking (on TRAIN only)
from sklearn.feature_selection import mutual_info_classif

remaining = [c for c in X_train_enc.columns if c not in to_drop_corr + low_var]
discrete_mask = np.array([c.startswith('cat__') for c in remaining])
mi = mutual_info_classif(X_train_enc[remaining], y_train,
                         discrete_features=discrete_mask, random_state=42)
mi_series = pd.Series(mi, index=remaining).sort_values(ascending=False)

plt.figure(figsize=(9, 8))
mi_series.head(25)[::-1].plot(kind='barh', color='#1a3c6e')
plt.title('Top 25 features by mutual information with price_category')
plt.tight_layout()
plt.savefig('feature_importance_mi.png', dpi=150)
plt.show()

MI_THRESHOLD = 0.005   # you can tune this after looking at the chart
core_prefixes = ('cat__room_type', 'cat__neighbourhood_group_cleansed')   # always keep the project's core features
selected_features = [c for c in remaining
                     if mi_series[c] > MI_THRESHOLD or c.startswith(core_prefixes)]

print(f"\nTotal encoded: {X_train_enc.shape[1]} | after filters: {len(remaining)} | selected: {len(selected_features)}")
```
